# UNIVARIATE ANALYSIS

### 1. Top 15 Companies by Layoff Records (Horizontal Gradient Bar Chart)
Used to compare values across multiple companies and easily identify the top contributors.

```python
plt.figure(figsize=(12,8))

company_counts = df['company_name'].value_counts().head(15)

sns.barplot(x=company_counts.values,y=company_counts.index,palette="rocket")
plt.title("Top 15 Companies Appearing in Workforce Dataset",fontsize=18,weight='bold')
plt.xlabel("Number of Records")
plt.ylabel("Company")

for i,v in enumerate(company_counts.values):
    plt.text(v+2,i,str(v),va='center',fontsize=10)

plt.grid(axis='x',alpha=0.3)
plt.show()
```

### KEY INSIGHTS
1. A small number of global technology companies dominate the dataset.
2. This indicates workforce changes are concentrated among industry leaders rather than evenly distributed.
3. Large firms undergo more frequent restructuring, hiring cycles, and public reporting.

### 2. Company Size Distribution (Donut Chart)
Used to show each industry's proportion of the whole dataset in percentage terms.

```python
industry = df["industry"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    industry,
    labels=industry.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.45)
)

plt.title("Industry Distribution")
plt.show()
```

### KEY INSIGHTS
1. Technology-focused industries occupy the largest share of the dataset.
2. AI, Cloud, and Software sectors dominate workforce changes, highlighting where restructuring is most active.

### 3. Distribution of Layoffs (Histogram + KDE)
Used to visualize the frequency distribution, spread, and skewness of layoff counts.

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["layoffs_count"],
    bins=35,
    kde=True,
    color="BLUE"
)

plt.title("Distribution of Layoffs Count",fontsize=15,fontweight="bold")
plt.xlabel("Layoffs")
plt.show()
```

### KEY INSIGHTS
1. The distribution is strongly right-skewed.
2. Most companies conduct relatively small layoffs, while a few organizations experience exceptionally large workforce reductions.

### 4. Trend of Total Layoffs Across Years (Line Chart)
Used to display trends and changes in layoffs over a time period.

```python
yearly = df.groupby("year")["layoffs_count"].sum().reset_index()

plt.figure(figsize=(10,5))

sns.lineplot(
    data=yearly,
    x="year",
    y="layoffs_count",
    marker="D",
    linewidth=3
)

plt.title("Year-wise Total Layoffs")
plt.xlabel("Year")
plt.ylabel("Layoffs")
plt.grid(alpha=.3)

plt.show()
```

### KEY INSIGHTS
1. Layoff activity changes noticeably across years rather than remaining stable.
2. Peaks indicate periods of economic uncertainty or organizational restructuring.

### 5. Correlation Heatmap (Numerical Variables)
Used to identify positive, negative, and weak correlations among numerical variables.

### 1. Industry vs Average Layoffs (Bar Plot)
A bar plot is used to compare the average layoff count across different industries and identify which industries are most affected.

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='industry',
    y='layoffs_count',
    data=df,
    estimator=np.mean,
    palette='viridis'
)

plt.xticks(rotation=45)
plt.title("Average Layoffs by Industry",fontsize=15,fontweight='bold')
plt.xlabel("Industry")
plt.ylabel("Average Layoffs")
plt.grid(axis='y',alpha=0.3)

plt.show()
```

### KEY INSIGHTS
1. The average layoff size differs substantially across industries.
2. Technology, Finance, and E-commerce sectors record the highest average layoffs, indicating greater workforce restructuring.
3. Traditional industries generally experience smaller workforce reductions.
4. This suggests that fast-changing industries are more sensitive to economic and technological disruption.

### 2. Company Size vs Layoffs (Box Plot)
A box plot is used to compare the distribution, spread, median, and outliers of layoffs across different company sizes.

```python
plt.figure(figsize=(9,6))

sns.boxplot(
    x='company_size',
    y='layoffs_count',
    data=df,
    palette='Set2'
)

plt.title("Layoff Distribution by Company Size",fontsize=15,fontweight='bold')
plt.xlabel("Company Size")
plt.ylabel("Layoffs Count")
plt.grid(alpha=0.3)

plt.show()
```

### KEY INSIGHTS
1. Large organizations exhibit the widest spread in layoff counts.
2. Startups generally have lower layoff volumes but show occasional extreme cases.
3. Enterprise companies carry higher workforce risk because even small restructuring decisions affect thousands of employees.
4. Workforce volatility increases with organizational scale

### 3. Revenue Growth vs Layoffs (Scatter Plot)
A scatter plot is used to examine whether revenue growth is related to the number of layoffs and identify any correlation between the two numerical variables.

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    x='revenue_growth_percent',
    y='layoffs_count',
    hue='industry',
    palette='tab10',
    s=90,
    alpha=0.8,
    data=df
)

plt.title("Revenue Growth vs Layoffs",fontsize=15,fontweight='bold')
plt.xlabel("Revenue Growth (%)")
plt.ylabel("Layoffs Count")

plt.show()
```

### KEY INSIGHTS

2. High-growth organizations tend to maintain smaller workforce reductions.
3. The relationship indicates that declining financial performance is a major driver of workforce optimization.
4. Industry-wise colors reveal that some sectors remain resilient despite revenue fluctuations.

### 1. AI Adoption vs Layoffs (Bubble Chart)
Why this chart is used: To visualize the relationship between four variables (AI adoption, layoffs, open roles, and market condition) simultaneously and identify hidden business patterns.

```python
fig = px.scatter(
    df,
    x="ai_adoption_level",
    y="layoffs_count",
    size="open_roles",
    color="employee_sentiment",
    hover_name="company_name",
    color_continuous_scale="Turbo",
    size_max=50,
    title="AI Adoption vs Layoffs"
)

fig.show()
```

2. Large bubbles indicate organizations continuing to hire despite workforce reductions.
3. Low employee sentiment is concentrated among high-layoff companies.
4. Companies balancing AI adoption with strong hiring tend to retain better sentiment.

### 2. Industry → Company Size → Market Condition (Treemap)
Why this chart is used: To compare the hierarchical contribution of industries, company sizes, and market conditions to total layoffs.

```python
fig = px.treemap(
    df,
    path=["industry","company_size","market_condition"],
    values="layoffs_count",
    color="layoffs_count",
    color_continuous_scale="RdYlBu_r"
)

fig.show()
```

### KEY INSIGHTS
1. Quickly identifies industries contributing the highest layoffs.
2. Reveals whether startups, medium, or enterprise companies dominate layoffs.
3. Highlights which market conditions coincide with workforce reductions.

Why this chart is used: To compare multiple workforce indicators across countries and identify regions with contrasting hiring and layoff trends.

### KEY INSIGHTS
1. Countries with strong revenue growth may still experience significant layoffs.

### ENCODING
Encoding converts text into numbers.

####  1. Label Encoding
Best for Ordered Categories (Ordinal Data)

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["company_size"] = le.fit_transform(df["company_size"])
df["company_size"]
```

#### 2. One-Hot Encoding 
Best for Nominal Data (no order).

```python
pd.get_dummies(
    df,
    columns=["industry"],
    dtype=int
)
```

#### 3. Ordinal Encoding

When categories have order.

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder()

df[["hiring_trend"]] = oe.fit_transform(df[["hiring_trend"]])
df[["hiring_trend"]]
```

### SCALING
Scaling makes numerical values comparable.

#### 1. StandardScaler

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

num_cols = [
    "layoffs_count",
    "layoff_percentage",
    "ai_automation_impact",
    "ai_replacement_risk",
    "open_roles",
    "remote_jobs_percentage",
    "stock_growth_percent",
    "revenue_growth_percent",
    "salary_budget_change",
    "ai_adoption_level",
    "employee_sentiment",
    "job_security_score"
]

df[num_cols] = scaler.fit_transform(df[num_cols])
df[num_cols]
```

#### 2. MinMaxScaler

Range

0 to 1

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[num_cols] = scaler.fit_transform(df[num_cols])
df[num_cols]
```

#### 3. RobustScaler

Best when data has many outliers.

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

df[num_cols] = scaler.fit_transform(df[num_cols])
df[num_cols]
```
